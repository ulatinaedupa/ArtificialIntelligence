# Student Marks

Problem Statement:

- 🎓🎓 Analytics Vidhya conducted an assessment test to hire a Data Scientist. The
candidates were evaluated on 5 different subjects A, B, C, D, and E. The marks of
5 different subjects out of 100 are given below. Help Analytics Vidhya to find out
the answers of the following questions.

```python
# student marks

student_marks = [
    ['Name',     ['A', 'B', 'C', 'D', 'E']],
    ['Ankit',    [41,   34,  45,  55,  63]],
    ['Aravind',  [42,   23,  34,  44,  53]],
    ['Lakshay',  [32,   23,  13,  54,  67]],
    ['Gyan',     [23,   82,  23,  63,  34]],
    ['Pranav',   [21,   23,  25,  56,  56]]
]
```

```python
student_marks_dict = dict(student_marks)
names = list(student_marks_dict.keys())[1:]
```

- 🙇 Who scored the highest marks in the subject B?
- 🙇 What is the average marks scored in the subject C?

```python
subject_b = [student_marks[0:][i][1][1] for i in range(1,len(student_marks))]
max_subject_b = max(subject_b)
index_subject_b = subject_b.index(max_subject_b)
name_subject_b = student_marks[index_subject_b + 1][0]
print('highest mark B ->', max_subject_b, '->', name_subject_b)
```

```python
subject_c = [student_marks[0:][i][1][2] for i in range(1,len(student_marks))]
avg_subject_c = sum(subject_c)/len(subject_c)
print('average mark C ->', avg_subject_c)
```

- 🙇 Who scored the highest percentage of marks?

```python
subjects = student_marks[0][1]
max_mark_ix_ = 0
max_mark_ = 0
for nsubject, subject in enumerate(subjects):
    mark = [student_marks[1:][exam][1][nsubject] for exam in range(len(subjects))]
    max_mark = max(mark)
    max_mark_ix = mark.index(max_mark)
    if max_mark > max_mark_:
        max_mark_ = max_mark
        max_mark_ix_ = max_mark_ix
    #print(f'Subject {subject} -> {names[max_mark_ix]} -> {max_mark}')

print()
print(f'{names[max_mark_ix_]}')
```

- 🙇 If considered only top-4 subjects of a candidate, then who scored the highest
percentage of marks?

```python
for i in range(1, len(student_marks)):
    mark = min(student_marks[i][1])
    mark_ix = student_marks[i][1].index(mark)
    student_marks[i][1][mark_ix] = 0
```
